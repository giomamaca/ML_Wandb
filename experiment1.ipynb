```python
import os
from src.data import get_loaders
from src.models import MLP
from src.utils import check_initial_loss, overfit_small_batch, check_gradients
from src.train import train

CSV = '../train.csv',
print('Using data:', CSV)

Model    = MLP
run_name = '01_mlp_baseline'
config   = {'lr':1e-3,'epochs':30,'optimizer':'adam'}
augment  = False

tr, va = get_loaders(CSV, batch_size=64, augment_train=False)
print("===== sanity =====")
check_initial_loss(Model(), va)
overfit_small_batch(Model(), tr, steps=200)
check_gradients(Model(), tr)

tr, va = get_loaders(CSV, batch_size=64, augment_train=augment)
print(f"\n===== train {run_name} =====")
train(Model(), tr, va, config, run_name=run_name)
```
